```python
def count(n):
    for i in range(n):
        yield i
```

```python
count(5)
```

```python
for num in count(5): # count fun is iterator ---> yield
    print(num)
```

```python
def square(a):
    return a * a
```

```python
square(5)
```

```python
for i in square(5):
    print(i)
```

```python
for i in range(1,5+1):
    print(i)
```

```python
def square_gen(n):
    for i in range(1,n+1):
        yield i*i 
```

```python
for num in square_gen(10):
    print(num)
```
