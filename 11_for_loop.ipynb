```python
number = 3
if number > 1:
    for i in range(2,number):
        if number % i == 0:
            print(number, 'is not prime')
            break
        else:
            print(number, 'is prime number')
else:
    print(number, 'is not prime numer')
```

```python
number = 17
if number == 2:
    print(number, 'Prime number')
elif number > 1:
    for i in range(2,number):
        if number % i == 0:
            print(number, 'is not prime')
        else:
            print(number, 'is prime number')
            break
else:
    print(number, 'is not prime numer')
```
