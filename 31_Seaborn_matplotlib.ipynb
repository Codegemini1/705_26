```python
sns.barplot(data=df1, x='day', y='total_bill')
plt.title('Number of customers per day')
plt.show()
```

```python
sns.lineplot(data=df1,x='size', y='total_bill')
plt.title('Bill vs Tip')
plt.show()
```

```python
sns.lineplot(data=df1,x='size', y='tip')
plt.title('Bill vs Tip')
plt.show()
```

```python
sns.histplot(df1['total_bill'], bins=15)
plt.title('Bill distribution')
plt.show()
```

```python
sns.boxenplot(x='day', y='total_bill', data=df1)
plt.title('Bill Boxplot')
plt.show()
```
