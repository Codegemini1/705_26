# which category is having maximum installs ?

```python
df.groupby(['Category'])['Installs'].sum()
```

```python
df_cat_installs = df.groupby(['Category'])['Installs'].sum().sort_values(ascending=False).reset_index()
```

```python
df_cat_installs
```

```python
df_cat_installs.head()
```

```python
df_cat_installs['Installs'] = df_cat_installs['Installs']/1000000
```

```python
df_cat_installs
```

```python
# Top 10 category installs
```

```python
plt.figure(figsize=(14,10))
sns.set_context('talk')
sns.set_style('darkgrid')

ax = sns.barplot(x='Installs', y='Category', data=df_cat_installs.head(10))
ax.set_xlabel('No of installs in billion')
ax.set_ylabel('Category')
ax.set_title('Top 10 category wise installs')
```
