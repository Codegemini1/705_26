# Kwargs

```python
def test3(**kwargs):
    return kwargs
```

```python
test3(a = 2, b= 5)
```

```python
# Fun data entry
def test4(**kwargs):
    return kwargs.keys(), kwargs.values()
```

```python
test4(Name= 'Python', Age = 30, Number = 39287347984)
```

```python
import pandas as pd
def test4(**kwargs):
    result = pd.DataFrame(kwargs)
    return result
```
