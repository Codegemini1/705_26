```python
try:
    a = int(input('Write int num'))
    b = int(input('Write int num'))
except:
    print('Written wrong type of input')

try:
    result = a/b
    print(result)
except:
    print('Not possible to devide')

try:
    result = a * b
    print(result)
except:
    print('Can mutiply a with b')
```

```python
def division(a,b):
    try:
        result = a/b
        print(result)
    except Exception as e:
        print(e)
```

```python
division(10,3)
```

```python
division(5,0)
```
