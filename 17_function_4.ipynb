List Comprehension

```python
nums = [1,2,3,4,5]
```

```python
[i*i for i in nums]
```

```python
[i for i in nums if i % 2 == 0]
```

```python
names = ['rahul', 'shreya', 'rock']
```

```python
[i.upper() for i in names]
```

```python
'rock'.upper()
```
